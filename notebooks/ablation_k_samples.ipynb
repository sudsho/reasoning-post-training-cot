# K-samples ablation for self-consistency

Reads outputs of `scripts/self_consistency_eval.sh` and plots accuracy vs K, per bench, per model tag.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

ab = Path('../benchmarks/sc_ablation')
ks = [1, 2, 4, 8, 16]

def series(bench):
    ys = []
    for k in ks:
        p = ab / f'{bench}.k{k}.json'
        ys.append(json.loads(p.read_text())['acc'] if p.exists() else None)
    return ys

arc = series('arc_c')
gsm = series('gsm8k')
print('arc-c', dict(zip(ks, arc)))
print('gsm8k', dict(zip(ks, gsm)))
```

```python
plt.figure()
if any(x is not None for x in arc):
    plt.plot(ks, arc, marker='o', label='ARC-Challenge')
if any(x is not None for x in gsm):
    plt.plot(ks, gsm, marker='s', label='GSM8K')
plt.xscale('log', base=2)
plt.xticks(ks, ks)
plt.xlabel('K samples')
plt.ylabel('accuracy')
plt.title('self-consistency: accuracy vs K')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
